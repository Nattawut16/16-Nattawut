```python
from datetime import datetime

current_year = datetime.now().year

birth_year = int(input("กรอกปีเกิด (ค.ศ.): "))

age = current_year - birth_year

print(f"ปีปัจจุบันคือ {current_year}")
print(f"คุณอายุประมาณ {age} ปี")
```
