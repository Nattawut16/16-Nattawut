```python
from datetime import datetime, timedelta, time

# ฟังก์ชันสำหรับจัดการเวลาไทย
def get_thai_time():
    # UTC+7 สำหรับเวลาไทย
    return datetime.utcnow() + timedelta(hours=7)

def get_valid_number(prompt):
    while True:
        try:
            value = float(input(prompt)) # ใช้ float เผื่อใส่ 1.5 ชม.
            if value < 0:
                print("ตัวเลขต้องมากกว่า 0 ครับ")
                continue
            return value
        except ValueError:
            print("⚠ กรุณาใส่ตัวเลขเท่านั้น")

# --- ส่วนแสดงผลเริ่มต้น ---
now_thai = get_thai_time()
print("="*40)
print(f"🇹🇭 ขณะนี้เวลาไทย: {now_thai.strftime('%H:%M')} น.")
print("="*40)

# --- รับข้อมูล ---
names1 = input("1. ชื่อกิจกรรมที่ 1 : ")
hour1 = get_valid_number("   จำนวนชั่วโมง   : ")

names2 = input("2. ชื่อกิจกรรมที่ 2 : ")
hour2 = get_valid_number("   จำนวนชั่วโมง   : ")

print("-" * 20)
# รับเวลาตื่นนอนแบบระบุเจาะจง (เช่น 06.00)
while True:
    try:
        wake_str = input("⏰ ต้องการตื่นกี่โมง (เช่น 06:00, 7.30): ").replace('.', ':')
        if ':' not in wake_str: # ถ้าใส่มาแค่เลข 6 ให้แก้เป็น 6:00
            wake_str += ":00"

        h, m = map(int, wake_str.split(':'))
        wake_up_time = time(h, m) # สร้าง object เวลา
        break
    except ValueError:
        print("⚠ รูปแบบเวลาไม่ถูกต้อง กรุณาลองใหม่ (เช่น 06:30)")

sleep_duration = get_valid_number("😴 ต้องการนอนกี่ชั่วโมง: ")

# --- การคำนวณ (Logic เวลาจริง) ---
# 1. คำนวณเวลานอน (Bedtime)
# สร้างเวลาตื่นของ "พรุ่งนี้" (สมมติ) เพื่อลบด้วยเวลานอน
dummy_date = datetime(2000, 1, 2, wake_up_time.hour, wake_up_time.minute)
bedtime_dt = dummy_date - timedelta(hours=sleep_duration)

# 2. คำนวณเวลากิจกรรมรวม
total_hours = hour1 + hour2

# --- แสดงผล ---
print("\n" + "📅 ตารางสรุปกิจกรรม (โซนเวลาไทย)".center(40, "="))
print(f"• {names1:<15} ใช้เวลา {hour1} ชม.")
print(f"• {names2:<15} ใช้เวลา {hour2} ชม.")
print(f"• เวลากิจกรรมรวม    : {total_hours} ชม.")
print("-" * 40)
print(f"🔔 เป้าหมาย: ตื่นตอน {wake_up_time.strftime('%H:%M')} น.")
print(f"🛌 เวลาที่ต้องเข้านอน : {bedtime_dt.strftime('%H:%M')} น.")
print("=" * 40)
```
